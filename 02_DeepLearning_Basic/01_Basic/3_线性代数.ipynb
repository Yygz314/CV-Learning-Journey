## 线性代数

标量由只有一个元素的张量表示

```python
import torch

x = torch.tensor([3.0])
y = torch.tensor([2.0])

x + y, x * y, x / y, x ** y 
```

可以将向量视为标量值组成的列表

```python
x = torch.arange(4)
x
```

通过张量的索引来访问任意元素

```python
x[3]
```

访问张量的长度

```python
len(x)
```

只有一个轴的张量，形状只有一个元素

```python
x.shape
```

指定通过两个分量m和n来创建一个m*n的矩阵

```python
A = torch.arange(20).reshape(5,4)
A
```

矩阵的转置

```python
A.T
```

对称矩阵 其转置等于自己

```python
B = torch.tensor([[1,2,3],[2,0,4],[3,4,5]])
B
```

```python
B == B.T
```

就像向量是标量的推广，矩阵是向量的推广。可以构建具有更多轴的数据结构

```python
x = torch.arange(24).reshape(2,3,4)
x
```

给定具有相同形状的任何两个张量，任何按元素二元运算的结果都将是相同形状的张量

```python
A = torch.arange(20,dtype=torch.float32).reshape(5,4)
B = A.clone()
A, A + B
```

两个矩阵的按元素乘法称为 哈达玛积 

```python
A * B
```

```python
a = 2
X = torch.arange(24).reshape(2,3,4)
a + X, (a * X).shape
```

计算元素的和

```python
x = torch.arange(4,dtype=torch.float32)
x, x.sum()
```

表示任意形状张量的元素和

```python
A = torch.arange(20).reshape(5,4)
A.shape, A.sum()
```

指定求和汇总张量的轴

```python
A_sum_axis0 = A.sum(axis = 0)
A_sum_axis0, A_sum_axis0.shape
```

```python
A_sum_axis1 = A.sum(axis=1)
A_sum_axis1, A_sum_axis1.shape
```

```python
A.sum(axis=[0,1]).shape
```

平均值

```python
A.mean(dtype=torch.float32), A.sum() / A.numel()
```

```python
A.mean(axis=0,dtype=torch.float32), A.sum(axis=0) / A.shape[0]
```

计算中和或均值时保持轴数不变

```python
sum_A = A.sum(axis = 1,keepdims = True)
sum_A
```

通过广播机制A除以sum_A

```python
A / sum_A
```

某个轴计算A元素的累积总和

```python
A.cumsum(axis = 0)
```
